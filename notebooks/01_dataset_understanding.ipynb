# Dataset Understanding

In this notebook, we load the PaySim dataset for the first time to understand its structure, columns, data types, and the class distribution of our target variable (`isFraud`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")
```

### Load Dataset
Ensure you have downloaded `PS_20174392719_1491204439457_log.csv` and placed it in the `data/raw/` directory.

```python
data_path = '../data/raw/PS_20174392719_1491204439457_log.csv'

try:
    df = pd.read_csv(data_path)
    print("Dataset loaded successfully!")
    print(f"Shape: {df.shape[0]:,} rows and {df.shape[1]} columns.")
except FileNotFoundError:
    print("Error: Dataset not found. Please download it from Kaggle and place it in 'data/raw/'.")
```

### Inspect Columns and Data Types

```python
if 'df' in locals():
    display(df.head())
    display(df.info())
```

### Target Variable Distribution (Class Imbalance)

```python
if 'df' in locals():
    fraud_counts = df['isFraud'].value_counts()
    fraud_percentage = fraud_counts / len(df) * 100
    
    print("Class Distribution:")
    print(fraud_counts)
    print(f"\nPercentage:\n{fraud_percentage}")
    
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x='isFraud')
    plt.title('Distribution of Fraudulent vs Legitimate Transactions')
    plt.yscale('log') # Log scale is useful because of massive imbalance
    plt.ylabel('Count (Log Scale)')
    plt.show()
```
